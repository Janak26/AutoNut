```python
import pymupdf as fitz 
import re
from typing import List
import json
import os
```

```python
def load_cookbook_for_dummies():
    doc = fitz.open("/home/linux/linux/ProjectsReal/chef/pdfs/Connie-Sarros-Students-Vegetarian-Cookbook-for-Dummies.pdf")
    pages = []
    for page in doc:
        page = page.get_text()
        pages.append(page)

    recepies = []
    general = []
    keywords = ["prep time", "cook time", "yield", "ingredients", "directions", "per serving", "tip", "vary it", "note"]
    mandatory_keywords = ["prep time", "ingredients", "directions", "per serving"]
    for page in doc:
        keyword_count = 0
        mandatory_keyword_count = 0
        page_text = page.get_text()
        page_text = page_text.lower()
        for kw in keywords:
            if kw in page_text:
                keyword_count = keyword_count + 1
        for mkw in mandatory_keywords:
            if mkw in page_text:
                mandatory_keyword_count = mandatory_keyword_count + 1
        if mandatory_keyword_count > 3:
            page_text = {'text': page_text}
            recepies.append(page_text)
        else:
            general.append(page_text)
    return recepies
```

```python
cookbook_for_dummies = load_cookbook_for_dummies()
```

```python
def load_recepies_collection():
    doc = fitz.open("/home/linux/linux/ProjectsReal/chef/pdfs/recepies_collection.pdf")
    recepies = []
    for i in doc:
        page = i.get_text(sort=True)
        page = {'text': page}
        recepies.append(page)
    return recepies
```

```python
def load_creative_vegeterian_collection():
    doc = fitz.open("/home/linux/linux/ProjectsReal/chef/pdfs/Creative Vegetarian Cooking.pdf")
    pages = []
    for page in doc:
        page = page.get_text(sort=True)
        pages.append(page)
    recepies = []
    general = []
    keywords = ["ingredients", "preparation method", "curry", "paste", "variation", "syrup"]
    mandatory_keywords = ["ingredients", "preparation method"]
    for page in doc:
        keyword_count = 0
        mandatory_keyword_count = 0
        page_text = page.get_text()
        page_text = page_text.lower()
        page_text = page_text.strip()
        for kw in keywords:
            if kw in page_text:
                keyword_count = keyword_count + 1
        for mkw in mandatory_keywords:
            if mkw in page_text:
                mandatory_keyword_count = mandatory_keyword_count + 1
        if mandatory_keyword_count > 1:
            page_text = {'text': page_text}
            recepies.append(page_text)
        else:
            general.append(page_text)
    return recepies
```

```python
def load_kens_recepies():
    folder = "/home/linux/linux/ProjectsReal/chef/pdfs/Ken/{}.pdf"
    recepies = []
    for number in range(1, 77):
        number = str(number)
        filename = "/home/linux/linux/ProjectsReal/chef/pdfs/Ken/{}.pdf".format(number)
        doc = fitz.open(filename)
        pages = []
        for i in doc:
            page = i.get_text(sort=True)
            pages.append(page)
        pages = ' '.join(pages)
        pages = {'text': pages}
        recepies.append(pages)
    
    
    doc = fitz.open("/home/linux/linux/ProjectsReal/chef/pdfs/Ken/Bold Vegetarian Chef - Ken Charney - Process.pdf")
    pages = []
    for page in doc:
        page = page.get_text(sort=True)
        page = {'text': page}
        pages.append(page)
    recepies = recepies + pages
    return recepies
```
