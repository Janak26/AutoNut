````python
api_key = ""
client = OpenAI(api_key=api_key)
````

````python
KNOWLEDGE_BASE = """
You are a chef. structure the recepie in a proper format from the given text.
# sometimes, there can be more than one recipie or dish

Arrange the below text in correct order and return the result.
    The json should have the following keys only:
    -recepie name (strictly a string)
    - prep time (stictly a string)
    - cook time (stictly a string)
    - yield (stictly a string)
    - ingredients (very very stictly a single list of sentences)
    - directions (very very stictly a single list of sentences)
    - per serving (stictly a string)
    - note (if present in the original text) (stictly a string)
    - vary it (if present in the original text) (stictly a string)
    - tip (if present in the original text) (stictly a string)
    - prep time minutes (total prep time in minutes, strictly an integer)
    - cook time minutes (total cook time in minutes, strictly an integer)
    - tags (strictly a list) (important inormation if the recepie is for a drink, food item, starter, lunch, dinner, snacks, sweet, spicy, cuisine type, and many more that you can think of. give between 5 to 8 tags)

    double check the data structure and return as asked
    Return JSON.
"""

def extract_timeline(text):
    response = client.chat.completions.create(
        model="gpt-4.1-mini",  # fast + cheap + good for extraction
        temperature=0,
        messages=[
            {
                "role": "system",
                "content": KNOWLEDGE_BASE
            },
            {
                "role": "user",
                "content": f"Input text:\n{text}"
            }
        ]
    )

    result = response.choices[0].message.content
    return result
    # if result:
        # return json.loads(result)
    # return None
````

````python
formatted = []
errored = []
````

````python
for ind, recepie in enumerate(data):
    if ind%25 == 0:
        print(ind)
    extracted_recepie = extract_timeline(recepie)
    try:
        formatted_recepie = parse_chatgpt_json(extracted_recepie)
        formatted.append(formatted_recepie)
    except:
        errored.append(extracted_recepie)
````

````python
post_formatting = []
for recepie in formatted:
    
    if type(recepie) == dict:
        post_formatting.append(recepie)
    if type(recepie) == list:
        temp_list = []
        for single_return in recepie:
            temp_list.append(single_return)
        post_formatting = post_formatting + temp_list
````
