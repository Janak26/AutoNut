```python
def dict_to_list(text):
    empty = []
    for k,v in text.items():
        empty.append(k)
        if type(v) == list:
            v = convert_to_str(v)
        empty.append(v)
    return empty



def convert_to_list(text): # "ingredients", "directions"
    if type(text) == dict:
        text = dict_to_list(text)
        return text
    if type(text) == list:
        return text
    text = str(text)
    text = text.split('.')
    return text

def convert_to_str(text): # "vary it", "tip", "note", "per serving", "cook time", "prep time", "yield", "recipie name", "id", "text"
    if type(text) == dict:
        text = dict_to_list(text)
    if type(text) == list:
        stringer = ' '.join(text)
        return stringer
    if type(text) == str:
        return text
    return ''

def convert_int(text): # "cook time minutes", "prep time minutes"
    if type(text) == int:
        return text
    return 0
```

```python
final_json = []
for idnum, document in enumerate(data):
    try:
        updated_doc = {}
        non_existing = []
        text = []
        doc_keys = list(document.keys())
        # if 'id' in doc_keys:
        updated_doc['id'] = idnum
            
        if 'recepie name' in doc_keys:
            var_recepie_name = convert_to_str(document['recepie name'])
            updated_doc['recepie name'] = var_recepie_name
            text.append(var_recepie_name)
        else:
            updated_doc['recepie name'] = ''
        
        if 'ingredients' in doc_keys:
            var_ingredients = convert_to_list(document['ingredients'])
            updated_doc['ingredients'] = var_ingredients
            ingredients_string = convert_to_str(var_ingredients)
            text.append('ingredients = ')
            text.append(ingredients_string)
        else:
            updated_doc['ingredients'] = []
    
        if 'directions' in doc_keys:
            var_directions = convert_to_list(document['directions'])
            updated_doc['directions'] = var_directions
            directions_string = convert_to_str(var_directions)
            text.append('directions = ')
            text.append(directions_string)
        else:
            updated_doc['directions'] = []

        if 'tags' in doc_keys:
            var_tags = convert_to_list(document['tags'])
            updated_doc['tags'] = var_tags
            tags_string = convert_to_str(var_tags)
            text.append('tags = ')
            text.append(tags_string)
        else:
            updated_doc['tags'] = []
    
        if 'prep time' in doc_keys:
            var_prep_time = convert_to_str(document['prep time'])
            updated_doc['prep time'] = var_prep_time
            if len(var_prep_time) > 2:
                text.append('prep time = ')
                text.append(var_prep_time)
        else:
            updated_doc['prep time'] = ''
    
        if 'cook time' in doc_keys:
            var_cook_time = convert_to_str(document['cook time'])
            updated_doc['cook time'] = var_cook_time
            if len(var_cook_time) > 2:
                text.append('cook time = ')
                text.append(var_cook_time)
        else:
            updated_doc['cook time'] = ''
    
    
        if 'yield' in doc_keys:
            var_yield = convert_to_str(document['yield'])
            updated_doc['yield'] = var_yield
            if len(var_yield) > 2:
                text.append('yield = ')
                text.append(var_yield)
        else:
            updated_doc['yield'] = ''
    
        if 'cook time minutes' in doc_keys:
            var_cook_time_minutes = convert_int(document['cook time minutes'])
            updated_doc['cook time minutes'] = var_cook_time_minutes
    
        if 'prep time minutes' in doc_keys:
            var_prep_time_minutes = convert_int(document['prep time minutes'])
            updated_doc['prep time minutes'] = var_cook_time_minutes
    
        if 'per serving' in doc_keys:
            var_per_serving = convert_to_str(document['per serving'])
            updated_doc['per serving'] = var_per_serving
            if len(var_per_serving) > 2:
                text.append('per serving = ')
                text.append(var_per_serving)
        else:
            updated_doc['per serving'] = ''
    
        if 'note' in doc_keys:
            var_note = convert_to_str(document['note'])
            updated_doc['note'] = var_note
            if len(var_note) > 2:
                text.append('note = ')
                text.append(var_note)
        else:
            updated_doc['note'] = ''
    
        if 'tip' in doc_keys:
            var_tip = convert_to_str(document['tip'])
            updated_doc['tip'] = var_tip
            if len(var_tip) > 2:
                text.append('tip = ')
                text.append(var_tip)
        else:
            updated_doc['tip'] = ''
    
        if 'vary it' in doc_keys:
            var_vary_it = convert_to_str(document['vary it'])
            updated_doc['vary it'] = var_vary_it
            if len(var_vary_it) > 2:
                text.append('variation = ')
                text.append(var_vary_it)
        else:
            updated_doc['vary it'] = ''
        updated_doc['text'] = ' '.join(text)
        final_json.append(updated_doc)
    except:
        print("missed one")
```
